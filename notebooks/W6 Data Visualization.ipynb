## Introduction

Data visualization is the graphical representation of data. A useful figure is more than attractive: it should make a specific comparison easy to see without distorting the evidence. In this tutorial, we use [Matplotlib](https://matplotlib.org) and [Seaborn](https://seaborn.pydata.org) to build common statistical graphics and discuss when each one is appropriate.

We use the words *chart*, *plot*, and *graph* interchangeably. The examples rely on datasets distributed through Seaborn, so the first run requires an internet connection unless those datasets are already cached.

## Start with the question

Choose the chart from the comparison you want the reader to make—not from a desire to use as many chart types as possible.

| Analytical question | Useful starting point |
|---|---|
| How does a value change over an ordered axis or time? | Line chart |
| How do categories compare? | Bar or dot plot |
| What is the shape and spread of a numeric variable? | Histogram or boxplot |
| How do two numeric variables relate? | Scatterplot |
| How does a relationship differ across groups? | Color/marker encoding or small multiples |
| Where are high and low values in a matrix? | Heatmap |

Every chart should have a clear title, labeled axes with units, and enough context to identify the population or period being shown.


## Line Chart

The line chart is one of the simplest and most widely used data visualization techniques. A line chart displays information as a series of data points or markers connected by straight lines. You can customize the shape, size, color, and other aesthetic elements of the lines and markers for better visual clarity.

Here's a Python list showing some monthly temperature values.

```python
temperature = [90, 92, 98, 95, 100, 102]
```

We can visualize how temperature changes over time using a line chart. To draw a line chart, we can use the `plt.plot` function.

```python
plt.plot(temperature)
```

Calling the `plt.plot` function draws the line chart as expected. 

Let's enhance this plot step-by-step to make it more informative and beautiful.

### Customizing the X-axis

The X-axis of the plot currently shows list element indexes 0 to 5. The plot would be more informative if we could display the month for which we're plotting the data. We can do this by two arguments `plt.plot`.

```python
months_number = [3, 4, 5, 6, 7, 8]
```

```python
plt.plot(months_number, temperature)
```

You can also use text as the x-axis ticks

```python
months_text = ["Mar","Apr","May","Jun","Jul","Aug"]

plt.plot(months_text, temperature)
```

### Axis Labels

We can add labels to the axes to show what each axis represents using the `plt.xlabel` and `plt.ylabel` methods.

```python
plt.plot(months_text, temperature)

plt.xlabel('Months')
plt.ylabel('Temperature (°F)')
```

### Plotting Multiple Lines

You can invoke the `plt.plot` function once for each line to plot multiple lines in the same graph. Let's compare the temperature between Tallahassee and NYC.

```python
temp_tlh = [90, 92, 98, 95, 100, 102]
temp_nyc = [88, 84, 91, 95, 96, 93]
```

```python
plt.plot(months_text, temp_tlh)
plt.plot(months_text, temp_nyc)

plt.xlabel('Months')
plt.ylabel('Temperature (°F)')
```

### Chart Title and Legend

To differentiate between multiple lines, we can include a legend within the graph using the `plt.legend` function. We can also set a title for the chart using the `plt.title` function.

```python
plt.plot(months_text, temp_tlh)
plt.plot(months_text, temp_nyc)

plt.xlabel('Months', fontsize=14) #
plt.ylabel('Temperature', fontsize=14) #add fontsize parameter


#Adding a title
plt.title("Monthly Mean Temperature",fontsize = 16)

#Adding a legend
plt.legend(['Tallahassee', 'New York City']);
```

### Line Markers

We can also show markers for the data points on each line using the `marker` argument of `plt.plot`. Matplotlib provides many different markers, like a circle, cross, square, diamond, etc. You can find the full list of marker types here: https://matplotlib.org/stable/api/markers_api.html.

```python
plt.plot(months_text, temp_tlh, marker="o")
plt.plot(months_text, temp_nyc, marker="x")
plt.xlabel('Months')
plt.ylabel('Temperature (°F)')

#Adding a title
plt.title("Monthly Mean Temperature")

#Adding a legend
plt.legend(['Tallahassee', 'New York City']);
```

### Styling Lines and Markers

The `plt.plot` function supports many arguments for styling lines and markers:

- `color` or `c`: Set the color of the line ([supported colors](https://matplotlib.org/stable/gallery/color/named_colors.html))
- `linestyle` or `ls`: Choose between a solid or dashed line
- `linewidth` or `lw`: Set the width of a line
- `markersize` or `ms`: Set the size of markers
- `markeredgecolor` or `mec`: Set the edge color for markers
- `markeredgewidth` or `mew`: Set the edge width for markers
- `markerfacecolor` or `mfc`: Set the fill color for markers
- `alpha`: Opacity of the plot


Check out the documentation for `plt.plot` to learn more: [https://matplotlib.org/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib.pyplot.plot](https://matplotlib.org/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib.pyplot.plot) .

```python
plt.plot(months_text, temp_tlh, marker='s', c='b', ls='-', lw=2, ms=8, mew=2, mec='navy')
plt.plot(months_text, temp_nyc, marker='o', c='r', ls='--', lw=3, ms=10, alpha=0.5)


plt.xlabel('Months')
plt.ylabel('Temperature (°F)')

#Adding a title
plt.title("Monthly Mean Temperature")

#Adding a legend
plt.legend(['Tallahassee', 'New York City']);
```

### Treat dates as dates

Text month labels are fine for a small introductory example, but real time series should normally use date values. This lets Matplotlib format and space the time axis correctly.


```python
monthly_dates = pd.date_range("2025-03-01", periods=6, freq="MS")

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(monthly_dates, temp_tlh, marker="o", label="Tallahassee")
ax.plot(monthly_dates, temp_nyc, marker="s", linestyle="--", label="New York City")
ax.set(
    title="Monthly Mean Temperature, March–August 2025",
    xlabel="Month",
    ylabel="Temperature (°F)",
)
ax.legend()
fig.autofmt_xdate()
plt.show()
```

## Scatter Plot

In a scatter plot, the values of 2 variables are plotted as points on a 2-dimensional grid. Additionally, you can also use a third variable to determine the size or color of the points. Let's try out an example.

Here we use a simple taxis dataset from the seaborn library as one of the example datasets.

```python
# Load data into a Pandas dataframe
taxi = sns.load_dataset("taxis")
```

```python
taxi.head()
```

Let's try to visualize the relationship between taxi trip distance and trip total cost. 

You can use `plt.scatter` or `sns.scatterplot` to make scatterplots

```python
plt.scatter(taxi.distance, taxi.total)
```

```python
sns.scatterplot(x=taxi.distance, y=taxi.total)
```

For both plots you can use the `color` parameter to control the color and `s` parameter to control the size of the dot.

```python
sns.scatterplot(x=taxi.distance, y=taxi.total,color='red',s=10)
```

### Plotting using Pandas Data Frames

Seaborn has in-built support for Pandas data frames. Instead of passing each column as a series, you can provide column names and use the `data` argument to specify a data frame.

```python
plt.title('Taxis Trip Distance vs. Cost',fontsize=14)

sns.scatterplot(x='distance', 
                y='total', 
                s=50,
                data=taxi);
```

### Encoding groups and reducing overplotting

Scatterplots can encode a third variable with color or marker style. Transparency (`alpha`) helps reveal dense regions where points overlap. When one chart becomes crowded, small multiples often provide a clearer comparison.

The view below is explicitly limited to trips of at most 30 miles and totals of at most $100 so that the dense part of the data is readable. In an analysis, always disclose filters like these.


```python
taxi_view = taxi.loc[
    taxi["distance"].between(0, 30)
    & taxi["total"].between(0, 100)
].dropna(subset=["pickup_borough", "payment"])

fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(
    data=taxi_view,
    x="distance",
    y="total",
    hue="pickup_borough",
    alpha=0.35,
    s=25,
    ax=ax,
)
ax.set(
    title="Taxi Trip Cost Increases with Distance",
    xlabel="Trip distance (miles)",
    ylabel="Total cost ($)",
)
ax.legend(title="Pickup borough", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.show()
```

Color is useful when there are only a few groups. Faceting separates groups into aligned panels and reduces repeated legend lookup.


```python
g = sns.relplot(
    data=taxi_view,
    x="distance",
    y="total",
    col="payment",
    hue="pickup_borough",
    alpha=0.35,
    s=20,
    height=4,
    aspect=1.1,
)
g.set_axis_labels("Trip distance (miles)", "Total cost ($)")
g.set_titles("Payment: {col_name}")
plt.suptitle("Taxi Cost and Distance by Payment Type", y=1.05)
plt.show()
```

## Histogram

A histogram represents the distribution of a variable by creating bins (interval) along the range of values and showing vertical bars to indicate the number of observations in each bin. 

For example, let's visualize the distribution of values of distance the taxi dataset. We can use the `plt.hist` function to create a histogram.

```python
taxi.distance
```

```python
plt.title("Distribution of Taxi Trip Distances")

plt.hist(taxi.distance)
```

We can see that the most trip distance is short, say, below 10 miles.

### Controlling the size and number of bins

We can control the number of bins or the size of each one using the bins argument.


```python
plt.hist(taxi.distance, bins=20)
```

```python
import numpy as np

# Specifying the boundaries of each bin
plt.hist(taxi.distance, bins=np.arange(0, 30, 2.5));
```

### Multiple Histograms

Similar to line charts, we can draw multiple histograms in a single chart. We can reduce each histogram's opacity so that one histogram's bars don't hide the others'.

Let's draw separate histograms for each pickup location in NYC.

```python
taxi.pickup_borough.unique()
```

```python
#manh_df = taxi[taxi.pickup_borough == 'Manhattan']

quee_df = taxi[taxi.pickup_borough == 'Queens']

bron_df = taxi[taxi.pickup_borough == 'Bronx']

broo_df = taxi[taxi.pickup_borough == 'Brooklyn']
```

```python
bins = np.arange(0, 30, 2.5)
#plt.hist(manh_df.distance, alpha=0.4, )
plt.hist(quee_df.distance, alpha=0.4, bins=bins)
plt.hist(bron_df.distance, alpha=0.4, bins=bins)
plt.hist(broo_df.distance, alpha=0.4, bins=bins)

plt.title('Distribution of Taxi Trip by Pick up Location')
plt.legend(['Queens', 'Bronx', 'Brooklyn']);
```

### Small multiples for comparing distributions

Overlapping filled histograms can hide one another and become difficult to read. Faceting uses the same bins and x-axis in separate panels, making shapes easier to compare.


```python
borough_order = ["Manhattan", "Queens", "Brooklyn", "Bronx"]

g = sns.displot(
    data=taxi.dropna(subset=["pickup_borough"]),
    x="distance",
    col="pickup_borough",
    col_order=borough_order,
    col_wrap=2,
    bins=np.arange(0, 32.5, 2.5),
    stat="density",
    common_norm=False,
    color="steelblue",
    height=3,
)
g.set_axis_labels("Trip distance (miles)", "Density")
g.set_titles("{col_name}")
plt.suptitle("Trip-Distance Distributions by Pickup Borough", y=1.03)
plt.show()
```

## Bar Chart

Bar charts are quite similar to line charts, i.e., they show a sequence of values. However, a bar is shown for each value, rather than points connected by lines. We can use the `plt.bar` function to draw a bar chart.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(months_text, temp_tlh, color="steelblue")
ax.set(
    title="Monthly Mean Temperature in Tallahassee",
    xlabel="Month",
    ylabel="Temperature (°F)",
)
plt.show()
```

### Counts versus aggregated values

A histogram counts observations inside numeric intervals. A count plot counts observations in categories. A bar plot, by contrast, displays values that were supplied directly or calculated for each category.


```python
pickup_order = taxi["pickup_borough"].value_counts().index

fig, ax = plt.subplots(figsize=(8, 4))
sns.countplot(
    data=taxi,
    y="pickup_borough",
    order=pickup_order,
    color="steelblue",
    ax=ax,
)
ax.set(
    title="Taxi Trips by Pickup Borough",
    xlabel="Number of trips",
    ylabel="Pickup borough",
)
plt.show()
```

### Bar plots with averages and confidence intervals

Let's look at another sample dataset included with Seaborn, called `tips`. The dataset contains information about the sex, time of day, total bill, and tip amount for customers visiting a restaurant over a week.

```python
tips_df = sns.load_dataset("tips");
```

```python
tips_df
```

We might want to draw a bar chart to visualize how the average bill amount varies across different days of the week. One way to do this would be to compute the day-wise averages and then use `plt.bar` (try it as an exercise).

However, since this is a very common use case, the Seaborn library provides a `barplot` function which can automatically compute averages.

```python
fig, ax = plt.subplots(figsize=(8, 4))
sns.barplot(
    data=tips_df,
    x="day",
    y="total_bill",
    errorbar=("ci", 95),
    color="steelblue",
    ax=ax,
)
ax.set(
    title="Mean Restaurant Bill by Day (95% Confidence Intervals)",
    xlabel="Day",
    ylabel="Mean total bill ($)",
)
plt.show()
```

By default, the lines crossing each bar are **95% confidence intervals for the estimated mean**. They describe uncertainty in the mean, not the spread of individual bills. Use `errorbar="sd"` to show one standard deviation or `errorbar=None` to suppress intervals.

We can also specify a `hue` argument to compare estimates side by side using a third feature, such as sex. Whenever you show intervals, identify what they represent.


```python
sns.barplot(x='day', y='total_bill', hue='sex', data=tips_df, errorbar = None)
```

```python
sns.barplot(x='day', y='total_bill', hue='sex', data=tips_df)
```

You can make the bars horizontal simply by switching the axes.

```python
sns.barplot(x='total_bill', y='day', hue='sex', data=tips_df)

plt.xlim(0, 40) # To provide a range for x or y axis
```

### Show the distribution, not only the mean

Means and confidence intervals do not show the shape of the observations. A boxplot summarizes the median and spread, while an overlaid strip plot shows the underlying bills. This makes skew, overlap, and possible outliers visible.


```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(
    data=tips_df,
    x="day",
    y="total_bill",
    color="lightblue",
    showfliers=False,
    ax=ax,
)
sns.stripplot(
    data=tips_df,
    x="day",
    y="total_bill",
    color="black",
    alpha=0.45,
    jitter=0.2,
    size=3,
    ax=ax,
)
ax.set(
    title="Distribution of Restaurant Bills by Day",
    xlabel="Day",
    ylabel="Total bill ($)",
)
plt.show()
```

## Heatmap

A heatmap is used to visualize 2-dimensional data like a matrix or a table using colors. The best way to understand it is by looking at an example. We'll use another sample dataset from Seaborn, called `flights`, to visualize monthly passenger footfall at an airport over 12 years.

```python
flights_raw = sns.load_dataset("flights")
```

```python
flights_raw.head()
```

Note: check what the `.pivot` function is doing here.

```python
flights_df = flights_raw.pivot(index="month", columns="year", values="passengers")
```

```python
flights_df
```

`flights_df` is a matrix with one row for each month and one column for each year. The values show the number of passengers (in thousands) that visited the airport in a specific month of a year. We can use the `sns.heatmap` function to visualize the footfall at the airport.

```python
plt.title("No. of Passengers (1000s)")
sns.heatmap(flights_df)
```

The brighter colors indicate a higher footfall at the airport. By looking at the graph, we can infer two things:

- The footfall at the airport in any given year tends to be the highest around July & August.
- The footfall at the airport in any given month tends to grow year by year.

We can also display the actual values in each block by specifying `annot=True` and using the `cmap` argument to change the color palette.

```python
fig, ax = plt.subplots(figsize=(11, 6))
sns.heatmap(
    flights_df,
    fmt="d",
    annot=True,
    cmap="Blues",
    cbar_kws={"label": "Passengers (thousands)"},
    ax=ax,
)
ax.set(
    title="Monthly Airline Passengers",
    xlabel="Year",
    ylabel="Month",
)
plt.show()
```

### Match the color palette to the data

Use a **sequential** palette for values that run from low to high, as above. Use a **diverging** palette when deviations on both sides of a meaningful midpoint matter. Here, zero separates passenger counts below and above each month's long-run mean.


```python
monthly_anomaly = flights_df.sub(flights_df.mean(axis="columns"), axis="index")

fig, ax = plt.subplots(figsize=(11, 5))
sns.heatmap(
    monthly_anomaly,
    cmap="vlag",
    center=0,
    cbar_kws={"label": "Difference from monthly mean (thousands)"},
    ax=ax,
)
ax.set(
    title="Airline Passengers Relative to Each Month's Long-Run Mean",
    xlabel="Year",
    ylabel="Month",
)
plt.show()
```

## A scale can change the story

Bar length is interpreted from a zero baseline, so truncating the vertical axis can greatly exaggerate small differences. The two panels below contain identical order counts. The left panel is visually dramatic because its baseline is hidden; the right panel preserves the full bar lengths.


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
orders = [90, 92, 98, 95, 100, 102]

fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharex=True)

for ax in axes:
    ax.bar(days, orders, color="steelblue")
    ax.set_xlabel("Day")

axes[0].set_ylim(88, 104)
axes[0].set_title("Misleading: truncated baseline")
axes[0].set_ylabel("Orders")

axes[1].set_ylim(0, 110)
axes[1].set_title("Honest bar baseline")

fig.suptitle("Daily Orders: Identical Data, Different Visual Impression")
fig.tight_layout()
plt.show()
```

## Visualization checklist

Before sharing a figure, ask:

1. Does the chart type match the analytical question?
2. Are the title, axes, units, population, and time period clear?
3. Does the scale preserve an honest visual comparison?
4. Could aggregation be hiding spread, outliers, or subgroup differences?
5. Are colors meaningful and distinguishable, including without perfect color vision?
6. Is the chart readable without repeatedly searching a legend?
7. Are filters, uncertainty intervals, and transformations explained?
